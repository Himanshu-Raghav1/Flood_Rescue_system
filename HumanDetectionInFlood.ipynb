```python
from ultralytics import YOLO
```

```python

model = YOLO("yolov8s.pt")

model.train(
    data="/content/drive/MyDrive/FloodDetection/dataset/detecting humans in floods.v1i.yolov8/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    project="/content/drive/MyDrive/FloodDetection",
    name="training_run"
)
```

```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/FloodDetection/training_run3/weights/best.pt")

metrics = model.val(
    data="/content/drive/MyDrive/FloodDetection/dataset/detecting humans in floods.v1i.yolov8/data.yaml",
    split="test"
)

print(metrics)
```

```python
from ultralytics import YOLO

model = YOLO("/content/drive/MyDrive/FloodDetection/training_run3/weights/best.pt")

results = model.predict(
    source="/content/drive/MyDrive/FloodDetection/dataset/detecting humans in floods.v1i.yolov8/test/images",
    conf=0.3,
    save=True
)
```
